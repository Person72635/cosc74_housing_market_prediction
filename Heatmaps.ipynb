```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("/redfin_data (1).csv")

print(df.columns.tolist())
```

```python
features = ['MEDIAN_LIST_PRICE', 'MEDIAN_LIST_PPSF', 'MEDIAN_DOM', 'INVENTORY', 'NEW_LISTINGS', 'HOMES_SOLD', 'PENDING_SALES', 'OFF_MARKET_IN_TWO_WEEKS']

feature_data = df[features].copy()
print(feature_data.dtypes)
```

```python
correlation_matrix = feature_data.corr(method="spearman")
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,          # Display correlation values
    fmt=".2f",           # Two decimal places
    cmap="coolwarm",     # Blue = negative, red = positive
    vmin=-1,
    vmax=1,
    center=0,
    square=True,
    linewidths=0.5
)

plt.title("Training Feature Correlations — Spearman")
plt.xticks(rotation=45, ha="right")
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```
