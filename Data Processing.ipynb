```python
import pandas as pd
```

```python
# load dfs
redfin_df = pd.read_csv("redfin_data.csv")
census_df = pd.read_csv("census_data.csv")
```

```python
# some redfin processing
redfin_df = redfin_df[redfin_df['PROPERTY_TYPE'] == "All Residential"]
redfin_df['REGION'] = redfin_df['REGION'].str.replace('Zip Code: ', '', regex=False)
redfin_df = redfin_df.drop(columns=['PROPERTY_TYPE', 'STATE_CODE', 'MEDIAN_SALE_PRICE', 'MEDIAN_PPSF', 'MONTHS_OF_SUPPLY', 'PRICE_DROPS', 'SOLD_ABOVE_LIST'])
```

```python
# some census processing
census_df['NAME'] = census_df['NAME'].str.replace('ZCTA5 ', '', regex=False)
census_df = census_df.drop(columns=['zip_code', 'pct_male', 'pct_female'])
```

```python
# create the merged df
merged_df = pd.merge(redfin_df, census_df, left_on='REGION', right_on='NAME', how='inner')
merged_df = merged_df.dropna()
merged_df = merged_df.drop(columns=['NAME'])
cols = ['AVG_SALE_TO_LIST'] + [col for col in merged_df.columns if col != 'AVG_SALE_TO_LIST']
merged_df = merged_df[cols]
```

```python
region_counts = merged_df['REGION'].value_counts()

region_counts[region_counts == 39]

# Get regions that have exactly 39 occurrences
regions_to_keep = region_counts[region_counts == 39].index

# Filter merged_df
merged_df = merged_df[merged_df['REGION'].isin(regions_to_keep)]
```

```python
# Count unique values in the 'REGION' column of merged_df
unique_regions_count = merged_df['REGION'].nunique()
print(f"Number of unique zip codes in REGION: {unique_regions_count}")
```

```python
# Get sorted unique values of PERIOD_BEGIN and PERIOD_END
unique_period_begin = sorted(merged_df['PERIOD_BEGIN'].dropna().unique())
unique_period_end = sorted(merged_df['PERIOD_END'].dropna().unique())

print("Unique PERIOD_BEGIN dates:", unique_period_begin)
print("\nUnique PERIOD_END dates:", unique_period_end)
```

```python
from google.colab import files

def save_df(df, file_name):
    df.to_csv(file_name, index=False)
    files.download(file_name)
```

```python
market_dem_df = merged_df.drop(columns=['REGION'])
market_zip_df = merged_df.drop(columns=['total_population',
       'median_household_income', 'poverty_rate', 'pct_bachelors_or_higher',
       'median_age', 'pct_white_non_hispanic', 'pct_black_non_hispanic',
       'pct_native_american_non_hispanic', 'pct_asian_non_hispanic',
       'pct_pacific_islander_non_hispanic', 'pct_other_race_non_hispanic',
       'pct_two_or_more_races_non_hispanic', 'pct_hispanic_or_latino',
       'sex_ratio'])
market_df = market_zip_df.drop(columns=['REGION'])
```

```python
save_df(merged_df, 'market_zip_dem.csv')
save_df(market_dem_df, 'market_dem.csv')
save_df(market_zip_df, 'market_zip.csv')
save_df(market_df, 'market.csv')
```
