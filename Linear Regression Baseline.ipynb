```python
import autograd.numpy as np
from autograd import grad
import matplotlib.pyplot as plt
%matplotlib inline
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
import pandas as pd
```

```python
df = pd.read_csv('/content/market (1).csv')

dates = pd.to_datetime(df.iloc[:, 1])
test_df = df.loc[dates >= pd.Timestamp("2025-09-01")].copy()
train_df = df.loc[dates < pd.Timestamp("2025-09-01")].copy()

x_train = train_df.iloc[:, 3:].values
y_train = train_df.iloc[:, 0].values
x_test = test_df.iloc[:, 3:].values
y_test = test_df.iloc[:, 0].values
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, root_mean_squared_error, root_mean_squared_log_error

linearModel = LinearRegression()
linearModel.fit(x_train, y_train)

y_pred = linearModel.predict(x_test)
mse = mean_squared_error(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)
rmse = root_mean_squared_error(y_test, y_pred)
rmsle = root_mean_squared_log_error(y_test, y_pred)
print("MSE: ", mse)
print("MAE: ", mae)
print("RMSE: ", rmse)
print("RMSLE: ", rmsle)
```

```python
plt.scatter(y_test, y_pred)
plt.plot([0.9, 1.1], [0.9, 1.1], "r--", label="actual = predicted")
plt.plot
plt.xlabel("Actual")
plt.ylabel("Predicted")
plt.legend()

plt.show()
```
