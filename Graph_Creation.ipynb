```python
import autograd.numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
import pandas as pd
from sklearn.metrics import mean_absolute_error, root_mean_squared_error, r2_score
```

```python
shap_df = pd.read_csv('/content/shap_market_dem (1).csv')
market_dem_x_test = pd.read_csv('/content/market_dem_X_test (1).csv')
market_dem_y_test = pd.read_csv('/content/market_dem_y_test.csv')
y_preds_market_dem = pd.read_csv('/content/y_preds_market_dem.csv')
```

```python
x_dom = market_dem_x_test['MEDIAN_DOM']
y_dom = shap_df['MEDIAN_DOM']
plt.scatter(x_dom, y_dom)
plt.xlabel('Median Days on Market')
plt.ylabel('SHAP value')
plt.title('SHAP values for Median Days on Market')
plt.show()
```

```python
x_ma = market_dem_x_test['median_age']
y_ma = shap_df['median_age']
plt.scatter(x_ma, y_ma)
plt.xlabel('Median Age')
plt.ylabel('SHAP value')
plt.title('SHAP values for Median Age')
plt.show()
```

```python
x_phl = market_dem_x_test['pct_hispanic_or_latino']
y_phl = shap_df['pct_hispanic_or_latino']
plt.scatter(x_phl, y_phl)
plt.xlabel('Percent Hispanic or Latino')
plt.ylabel('SHAP value')
plt.title('SHAP values for Percent Hispanic or Latino')
plt.show()
```

```python
y_pred = y_preds_market_dem['0']
y_test = market_dem_y_test['AVG_SALE_TO_LIST']
ME = np.mean(y_test - y_pred)
mae = mean_absolute_error(y_test, y_pred)
rmse = root_mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
print("MAE: ", mae)
print("RMSE: ", rmse)
print("R2: ", r2)
print("ME: ", ME)
plt.scatter(y_test, y_pred)
plt.plot([0.9, 1.1], [0.9, 1.1], "r--", label="actual = predicted")
plt.plot
plt.xlabel("Actual Average Sale to List Price")
plt.ylabel("Predicted Average Sale to List Price")
plt.title("Actual vs Predicted Average Sale to List Price Ratio")
plt.legend()
```
